🌧️ Precipitation Cleaning Breakdown
| Part | What it does |
|---|---|
| `p, rh, rad = ...` | Creates shortcuts for column names |
| `describe()` | Shows rainfall statistics: mean, std, percentiles, max |
| `df[p] > 5, 10, 20...` | Counts how many rainfall values exceed each threshold |
| `big = ... > 30` | Collects unusually high rainfall values for investigation |
| `sort_values().head(30)` | Shows the 30 highest rainfall values |
| `value_counts()` | Checks if one station produces most extreme values |
| `groupby(date)` | Checks if extreme values are concentrated on one day |
| `±1 hour` | Checks rainfall before and after the suspicious value |
| `all stations at timestamp` | Checks whether other stations also recorded rain at that time |
| `shift(1), shift(-1)` | Gets the previous and next rainfall values |
| `neighbour_max` | Finds the maximum rainfall among the previous/next values |
| `share_wet` | Percentage of stations recording rain at the same time |
| `isolated` | Detects suspicious spikes: `>30 mm` + neighbors `<1 mm` + `<10%` stations raining |
| `precip_suspect` | Flags all rainfall values above 30 mm |
| `precip_spike_removed` | Flags spikes identified as suspicious |
| `np.nan` | Removes suspicious values by marking them as missing |
| `quantile(0.9999)` | Finds the 99.99th percentile as an upper limit |
| `clip()` | Limits extreme values to that upper limit |
| `log1p()` | Compresses large values to reduce their impact on the model |
| `rain_flag` | Converts rainfall into binary: `0 = no rain`, `1 = rain` |

```python
# Clear-sky index
df['clear_sky_index'] = np.where(
    df['EO_radiation'] > 0,
    df['radiation (W/m2)'] / df['EO_radiation'],
    np.nan
)

df['clear_sky_index'] = df[
    'clear_sky_index'
].clip(0, 1.2)


# Solar position
solar_position = pvlib.solarposition.get_solarposition(
    time=df['timestamp'],
    latitude=df['latitude'],
    longitude=df['longitude']
)

df['solar_zenith'] = solar_position['zenith'].values

df['is_daylight'] = (
    df['solar_zenith'] < 90
).astype(int)

g = df.groupby('station_id', group_keys=False)

# Rolling clear-sky index
df['csi_roll_1h'] = (
    df.groupby('station_id')
      .rolling('1h', on='timestamp')['clear_sky_index']
      .mean()
      .reset_index(level=0, drop=True)
      .reset_index(drop=True)
)

df['csi_roll_3h'] = (
    df.groupby('station_id')
      .rolling('3h', on='timestamp')['clear_sky_index']
      .mean()
      .reset_index(level=0, drop=True)
      .reset_index(drop=True)
)
```

## 5. THERMODYNAMIC FEATURES


```python
T = df['temperature (degrees Celsius)']
RH = df['relativehumidity (-)'].clip(0.01, 1)

# RH is assumed to be between 0 and 1
RH_percent = RH * 100

# Magnus formula
a = 17.62
b = 243.12

gamma = (
    np.log(RH_percent / 100)
    + (a * T) / (b + T)
)

df['dew_point'] = (
    b * gamma / (a - gamma)
)

# Dew point depression
df['dew_point_depression'] = T - df['dew_point']


# Vapor pressure (hPa)
df['vapor_pressure'] = (
    6.112
    * np.exp((17.67 * T) / (T + 243.5))
    * RH
)


# Temperature normalized to sea level
df['temperature_sea_level'] = (
    T + 0.0065 * df['elevation']
)
```

## 8. ROLLING WEATHER FEATURES

```python
rolling_windows = {
    '1h': 4,
    '6h': 24,
    '24h': 96
}

weather_columns = {
    'temperature (degrees Celsius)': 'temp',
    'relativehumidity (-)': 'rh',
    'radiation (W/m2)': 'radiation'
}

df = df.sort_values(
    ['station_id', 'timestamp']
).reset_index(drop=True)


for col, prefix in weather_columns.items():

    for name, window in rolling_windows.items():

        grouped = df.groupby('station_id')[col]

        df[f'{prefix}_roll_{name}_mean'] = (
            grouped
            .transform(
                lambda x: x.rolling(
                    window,
                    min_periods=1
                ).mean()
            )
        )

        df[f'{prefix}_roll_{name}_std'] = (
            grouped
            .transform(
                lambda x: x.rolling(
                    window,
                    min_periods=1
                ).std()
            )
        )

        df[f'{prefix}_roll_{name}_min'] = (
            grouped
            .transform(
                lambda x: x.rolling(
                    window,
                    min_periods=1
                ).min()
            )
        )

        df[f'{prefix}_roll_{name}_max'] = (
            grouped
            .transform(
                lambda x: x.rolling(
                    window,
                    min_periods=1
                ).max()
            )
        )
```

## 9. PRECIPITATION FEATURES

## 13. SPATIAL FEATURES

```python
stations = (
    df[
        ['station_id', 'latitude', 'longitude']
    ]
    .drop_duplicates('station_id')
    .set_index('station_id')
)


# Haversine distance between stations
def haversine(lat1, lon1, lat2, lon2):

    R = 6371.0

    lat1 = np.radians(lat1)
    lon1 = np.radians(lon1)

    lat2 = np.radians(lat2)
    lon2 = np.radians(lon2)

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        np.sin(dlat / 2) ** 2
        + np.cos(lat1)
        * np.cos(lat2)
        * np.sin(dlon / 2) ** 2
    )

    return 2 * R * np.arcsin(np.sqrt(a))


# Build nearest-station map
station_ids = stations.index.tolist()

nearest_stations = {}

for station in station_ids:

    lat1 = stations.loc[station, 'latitude']
    lon1 = stations.loc[station, 'longitude']

    distances = []

    for other in station_ids:

        if station == other:
            continue

        lat2 = stations.loc[other, 'latitude']
        lon2 = stations.loc[other, 'longitude']

        distance = haversine(
            lat1, lon1,
            lat2, lon2
        )

        distances.append((other, distance))

    distances.sort(key=lambda x: x[1])

    # k = 5 nearest stations
    nearest_stations[station] = [
        x[0] for x in distances[:5]
    ]
```

## 14. SAME-TIMESTAMP SPATIAL FEATURES

```python
# Create lookup tables
weather_at_time = df.set_index(
    ['timestamp', 'station_id']
)[
    [
        'temperature (degrees Celsius)',
        'relativehumidity (-)',
        'clear_sky_index'
    ]
]


def spatial_features(row):

    station = row['station_id']
    timestamp = row['timestamp']

    neighbors = nearest_stations.get(station, [])

    values = []

    for neighbor in neighbors:

        try:

            values.append(
                weather_at_time.loc[
                    (timestamp, neighbor)
                ]
            )

        except KeyError:
            pass

    if not values:

        return pd.Series({
            'neighbor_temp_mean': np.nan,
            'neighbor_rh_mean': np.nan,
            'neighbor_csi_mean': np.nan
        })

    values = pd.DataFrame(values)

    return pd.Series({
        'neighbor_temp_mean':
            values['temperature (degrees Celsius)'].mean(),

        'neighbor_rh_mean':
            values['relativehumidity (-)'].mean(),

        'neighbor_csi_mean':
            values['clear_sky_index'].mean()
    })


spatial = df.apply(
    spatial_features,
    axis=1
)

df = pd.concat(
    [df, spatial],
    axis=1
)


# Difference between station and neighbors
df['temp_neighbor_diff'] = (
    df['temperature (degrees Celsius)']
    - df['neighbor_temp_mean']
)

df['rh_neighbor_diff'] = (
    df['relativehumidity (-)']
    - df['neighbor_rh_mean']
)

df['csi_neighbor_diff'] = (
    df['clear_sky_index']
    - df['neighbor_csi_mean']
)
```
